### Preliminary Data Exploration

```python
def explore_dataframe(df, id_column=None):
    """
    Performs initial data exploration steps on a DataFrame.

    Args:
        df (pd.DataFrame): The input DataFrame.
        id_column (str, optional): The name of the column to check for duplicate IDs.
                                   Defaults to None.

    Returns:
        dict: A dictionary containing exploration results.
    """
    results = {}

    print("--- Data Exploration Report ---")
    print(f"DataFrame Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")

    # 1. Control sparsity of columns (percentage of missing values)
    print("1. Column Sparsity (Missing Values):")
    missing_values = df.isnull().sum()
    missing_percentage = (missing_values / len(df)) * 100
    sparsity_df = pd.DataFrame({
        'Missing Count': missing_values,
        'Missing Percentage': missing_percentage
    })
    sparsity_df = sparsity_df[sparsity_df['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)
    if not sparsity_df.empty:
        print(sparsity_df)
    else:
        print("No missing values found in any column.")
    results['column_sparsity'] = sparsity_df
    print("\n")

    # 2. Metrics of central tendency for numerical columns
    print("2. Central Tendency Metrics (Numerical Columns):")
    numerical_cols = df.select_dtypes(include=['number']).columns
    if not numerical_cols.empty:
        # Define a lambda for mode to ensure it returns a scalar or NaN
        # pd.Series.mode can return multiple modes or an empty series, which causes issues with .agg()
        def first_mode(x):
            mode_val = x.mode()
            if not mode_val.empty:
                return mode_val.iloc[0]
            return pd.NA # Use pd.NA for missing values consistently

        central_tendency = df[numerical_cols].agg(['mean', 'median', first_mode]).transpose()
        # Rename the 'first_mode' column to 'mode' for clarity
        central_tendency = central_tendency.rename(columns={'first_mode': 'mode'})
        print(central_tendency)
    else:
        print("No numerical columns found to calculate central tendency.")
    results['central_tendency'] = central_tendency if not numerical_cols.empty else "No numerical columns"
    print("\n")

    # 3. Check how many incomplete rows (rows with any missing values)
    print("3. Incomplete Rows:")
    incomplete_rows_count = df.isnull().any(axis=1).sum()
    incomplete_rows_percentage = (incomplete_rows_count / len(df)) * 100
    print(f"Total incomplete rows: {incomplete_rows_count} ({incomplete_rows_percentage:.2f}% of total rows)")
    results['incomplete_rows'] = {'count': incomplete_rows_count, 'percentage': incomplete_rows_percentage}
    print("\n")

    # 4. Check how many duplicated IDs specified by a column name
    if id_column and id_column in df.columns:
        print(f"4. Duplicated IDs in column '{id_column}':")
        duplicated_ids = df[df.duplicated(subset=[id_column], keep=False)].sort_values(by=id_column)
        num_duplicated_entries = duplicated_ids.shape[0]
        num_unique_duplicated_ids = duplicated_ids[id_column].nunique()

        if num_duplicated_entries > 0:
            print(f"Total entries with duplicated '{id_column}': {num_duplicated_entries}")
            print(f"Number of unique '{id_column}' values that are duplicated: {num_unique_duplicated_ids}")
            # Optionally, show the duplicated IDs
            # print("First 10 duplicated IDs (and their associated rows):")
            # print(duplicated_ids.head(10))
            # results['duplicated_ids_data'] = duplicated_ids
        else:
            print(f"No duplicated IDs found in column '{id_column}'.")
        results['duplicated_ids_summary'] = {
            'total_entries_with_duplicates': num_duplicated_entries,
            'unique_duplicated_ids_count': num_unique_duplicated_ids
        }
    elif id_column:
        print(f"Warning: '{id_column}' not found in DataFrame for duplicate check.\n")
        results['duplicated_ids_summary'] = "Column not found"
    else:
        print("4. No 'id_column' provided for duplicate ID check.\n")
        results['duplicated_ids_summary'] = "No ID column provided"

    print("--- Exploration Report End ---")
    return results
```

```python
import re

def standardize_column_names(df):
    """
    Standardizes column names of a DataFrame by:
    - Converting to lowercase.
    - Replacing spaces and hyphens with underscores.
    - Removing special characters.
    - Stripping leading/trailing underscores.

    Args:
        df (pd.DataFrame): The input DataFrame.

    Returns:
        pd.DataFrame: The DataFrame with standardized column names.
    """
    new_columns = []
    for col in df.columns:
        # Convert to lowercase
        col = str(col).lower()
        # Replace spaces and hyphens with underscores
        col = re.sub(r'[\s-]+', '_', col)
        # Remove any non-alphanumeric (except underscore) characters
        col = re.sub(r'[^a-z0-9_]', '', col)
        # Remove leading/trailing underscores
        col = col.strip('_')
        new_columns.append(col)
    df.columns = new_columns
    return df
```

```python
def count_empty_rows(df):
    """
    Count rows that contain any empty cells (NaN or blank strings).
    Returns dict: {'count': int, 'percentage': float}.
    """
    initial_row_count = df.shape[0]
    if initial_row_count == 0:
        return {'count': 0, 'percentage': 0.0}

    # Rimuove le righe con valori mancanti
    post_drop_df = df.dropna(inplace=False)

    # Salva il numero finale di righe
    final_row_count = post_drop_df.shape[0]
    count = initial_row_count - final_row_count
    
    return {'count': count, 'percentage': float(count / initial_row_count * 100)}

def drop_empty_rows(df):
    """
    Drop rows that contain any empty cells (NaN or blank strings).
    """
    initial_row_count = df.shape[0]

    df.dropna(inplace=True)

    dropped = initial_row_count - df.shape[0]

    print(f"Dropped {dropped} rows.")
```

```python
def duplicate_id_check(df, id_column):
    """
    Check for duplicate IDs in a specified column of a DataFrame.
    Returns dict: {'total_entries_with_duplicates': int, 'unique_duplicated_ids_count': int}.
    """
    if id_column not in df.columns:
        print(f"Warning: '{id_column}' not found in DataFrame for duplicate check.")
        return {'total_entries_with_duplicates': 0, 'unique_duplicated_ids_count': 0}

    duplicated_ids = df[df.duplicated(subset=[id_column], keep=False)]
    num_duplicated_entries = duplicated_ids.shape[0]
    num_unique_duplicated_ids = duplicated_ids[id_column].nunique()
    distinct_values_count = df[id_column].nunique()

    return {
        'total_entries_with_duplicates': num_duplicated_entries,
        'unique_duplicated_ids_count': num_unique_duplicated_ids,
        'distinct_values_count': distinct_values_count,
        'distinct_values_percentage': (distinct_values_count / df.shape[0]) * 100 if df.shape[0] > 0 else 0
    }
```

```python
def check_consistent_duplicates(df, id_columns, check_columns):
    """
    Handles duplicate checking using a single or composite key (e.g., title + artist).
    Verifies whether duplicates are consistent with respect to the columns specified in check_columns.
    
    Args:
        df (pd.DataFrame): The input DataFrame.
        id_columns (str or list): The column or list of columns that define uniqueness.
        check_columns (str or list): The column or list of columns to verify for consistency.
    """
    # Converte in lista se viene passata una stringa singola
    if isinstance(id_columns, str):
        id_columns = [id_columns]
        
    # Verifica presenza colonne di identificazione
    for col in id_columns:
        if col not in df.columns:
            print(f"Error: ID column '{col}' not found in DataFrame.")
            return pd.DataFrame()

    if not all(col in df.columns for col in check_columns):
        missing_cols = [col for col in check_columns if col not in df.columns]
        print(f"Error: The following check_columns were not found in DataFrame: {missing_cols}")
        return pd.DataFrame()

    # Trova i duplicati basandosi sulla combinazione delle id_columns (es. nome + artista)
    id_duplicates = df[df.duplicated(subset=id_columns, keep=False)].copy()

    if id_duplicates.empty:
        print(f"No duplicates found for {id_columns}. All rows are unique.")
        return pd.DataFrame()

    inconsistent_groups = []

    # Raggruppa per la chiave singola o composta
    for id_val, group in id_duplicates.groupby(id_columns):
        is_consistent = True
        for col in check_columns:
            if group[col].nunique() > 1:
                is_consistent = False
                break

        if not is_consistent:
            inconsistent_groups.append(group)

    if inconsistent_groups:
        print(f"Found {len(inconsistent_groups)} groups with inconsistent duplicates.")
        return pd.concat(inconsistent_groups).sort_values(by=id_columns).reset_index(drop=True)
    else:
        print(f"All duplicates are consistent across {check_columns}.")
        return pd.DataFrame()
```

```python
def drop_duplicate_rows_by_columns(df, match_columns, keep='first'):
    """
    Drop rows that are duplicated on all `match_columns`.
    Rows with the same key but different values in `match_columns` are kept.

    Args:
        df (pd.DataFrame): Input DataFrame.
        match_columns (list): Columns that must also match to consider a row duplicate.
        keep (str): Which duplicate to keep ('first', 'last', or False).

    Returns:
        pd.DataFrame: Cleaned DataFrame.
    """

    for col in match_columns:
        if col not in df.columns:
            print(f"Error: Column '{col}' not found in DataFrame.")

    initial_row_count = df.shape[0]

    df.drop_duplicates(subset=match_columns, keep=keep, inplace=True)

    final_row_count = df.shape[0]
    dropped_rows = initial_row_count - final_row_count

    print(f"Dropped {dropped_rows} duplicate rows.")
```

```python
def check_cross_dataset_consistency(df1, df2, id_col1, id_col2, compare_cols1, compare_cols2, dataset1_name, dataset2_name, common_ids):
    """
    Checks for consistency of descriptive columns for tracks common between two DataFrames.

    Args:
        df1 (pd.DataFrame): The first DataFrame.
        df2 (pd.DataFrame): The second DataFrame.
        id_col1 (str): The ID column name in df1.
        id_col2 (str): The ID column name in df2.
        compare_cols1 (list): List of descriptive columns in df1 to compare.
        compare_cols2 (list): List of descriptive columns in df2 to compare (must align with compare_cols1).
        dataset1_name (str): Name of the first dataset for printing.
        dataset2_name (str): Name of the second dataset for printing.
        common_ids (set): Set of common track IDs between the two datasets.

    Returns:
        bool: True if consistent, False if any inconsistency is found.
    """
    print(f"\n--- Comparing common tracks between {dataset1_name} and {dataset2_name} ---")
    if not common_ids:
        print(f"No common tracks found between {dataset1_name} and {dataset2_name}.")
        return True

    inconsistencies_found = False
    inconsistent_tracks_summary = []

    for track_id in list(common_ids):
        # Get all entries for the current track_id from both dataframes
        df1_entries = df1[df1[id_col1] == track_id]
        df2_entries = df2[df2[id_col2] == track_id]

        # Iterate through all combinations of entries for a given track_id
        for idx1, row1 in df1_entries.iterrows():
            for idx2, row2 in df2_entries.iterrows():
                for i, (col1, col2) in enumerate(zip(compare_cols1, compare_cols2)):
                    val1 = str(row1[col1]).strip().lower() if pd.notna(row1[col1]) else ''
                    val2 = str(row2[col2]).strip().lower() if pd.notna(row2[col2]) else ''

                    if val1 != val2:
                        inconsistencies_found = True
                        inconsistent_tracks_summary.append({
                            'track_id': track_id,
                            'dataset1_col': col1,
                            'dataset1_value': row1[col1],
                            'dataset2_col': col2,
                            'dataset2_value': row2[col2]
                        })
                        # Break from inner loop if one inconsistency is found for this track_id and pair of columns
                        break
                if not inconsistencies_found: # If no inconsistency found for this pair of entries, continue
                    continue
                else: # If inconsistency found, break outer loop to move to next track_id
                    break
            if inconsistencies_found: # If inconsistency found, break outer loop to move to next track_id
                break

    if inconsistencies_found:
        print(f"Inconsistencies found for {len(inconsistent_tracks_summary)} entries with common IDs.")
        print("First 5 inconsistencies:")
        display(pd.DataFrame(inconsistent_tracks_summary).head())
        return False
    else:
        print("All common tracks have consistent descriptive information across the datasets.")
        return True
```

```python
common_cols = (
    set(spotify_df1.columns)
    .intersection(spotify_df2.columns)
    - {"id"}
)

for col in common_cols:
    merged[col] = merged[f"{col}_df2"].combine_first(
        merged[f"{col}_df1"]
    )

    merged.drop(
        columns=[f"{col}_df1", f"{col}_df2"],
        inplace=True
    )

merged.drop(columns=['macro_subgenre', 'subgenre'], inplace=True)
```
